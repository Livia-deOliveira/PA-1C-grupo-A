```python
# Escrever um programa onde o usuário digita um número de 1 a 10 e o programa imprime a tabuada do número digitado.

num = int(input("Digite um número de 1 a 10: "))
for i in range (1, 11):
  print (num, "x", i, "=", num*i)
```

```python
# Escrever um programa onde o usuário digita um número e o programa imprime o número digitado do 0 até o número digitado.

num = int(input("Digite um número: "))
for i in range (0, num+1):
  print (i)
```

```python
# Escrever um programa onde o usuário digita um número e o programa imprime o número até o 0 (contagem regressiva).

num = int(input("Digite um número: "))
for i in range (num, -1, -1):
  print (i)
```

```python
# Escrever um programa onde o usuário digita um número e o programa digita os números pares do 0 até o número digitado.

num = int(input("Digite um número: "))
for i in range (0, num+1, 2):
  print (i)
```

```python
# Escrever um programa onde o usuário digita um número e o programa imprime os números impares do 0 até o número digitado.

num = int(input("Digite um número: "))
for i in range (1, num+1, 2):
  print (i)
```

```python
# Escrever um programa onde o usuário digita um número e o programa soma todos os números do 0 até o número digitado.

num = int(input("Digite um número: "))
soma = 0
for i in range (0, num+1):
  soma = soma + i
print (soma)
```

```python
# Escrever um programa onde o usuário digita uma palavra e o programa conta as letras da palavra digitada.

palavra = input("Digite uma palavra: ")
contador = 0
for i in palavra:
  contador = contador + 1
print (contador)
```

```python
# Escrever um programa onde o usuário digita uma palavra e o programa imprime a palavra invertida.

palavra = input("Digite uma palavra: ")
for i in range (len(palavra)-1, -1, -1):
  print (palavra[i], end="")
```

```python
# Escrever um programa onde o usuário digita 04 números e o programa imprime o maior.

num1 = int(input("Digite um número: "))
num2 = int(input("Digite um número: "))
num3 = int(input("Digite um número: "))
num4 = int(input("Digite um número: "))
maior = num1
if num2 > maior:
  maior = num2
if num3 > maior:
  maior = num3
if num4 > maior:
  maior = num4
print (maior)
```

```python
# Escrever um programa que recebe um número N e imprime uma pirâmide perfeitamente centralizada formada pelo número digitado.

num = int(input("Digite um número: "))
for i in range (1, num+1):
  print (" "*(num-i), end="")
  print ("*"*(2*i-1))
```
