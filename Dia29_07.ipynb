```python
# Escrever um programa que identifica se um número é positivo ou negativo.

numero = int(input("Digite um numero: "))

if numero >= 0:
    print("O número é positivo.")
else:
    print("O número é negativo.")
```

```python
# Escrever um programa que recebe um numero menor que 20 e imprime os numeros pares de 0 até o numero recebido

numero = int(input("Digite um número positivo menor que 20: "))

while numero > 20 or numero < 0:
    print("O número digitado é inválido, tente novamente.")
    numero = int(input("Digite um número positivo menor que 20: "))

if numero <= 20 and numero > 0:
    print("Os números pares de 0 até o número digitado são:")
    for i in range(0, numero +1, 2):
        print(i)
```
