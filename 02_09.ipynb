```python
# Exercício 4- Soma e Média.

notas = [7.5, 8.0, 6.5, 9.0, 8.5]

print ("Lista original: ")
print (notas)
print (" ")
soma = float(notas[0]) + float(notas[1]) + float(notas[2]) + float(notas[3]) + float(notas[4])
print ("Soma das notas: ")
print (soma)
print (" ")
media = soma / 5
print ("Média das notas: ")
print (media)
print (" ")
maior_nota = max(notas)
print ("Maior nota: ")
print (maior_nota)
print (" ")
menor_nota = min(notas)
print ("Menor nota: ")
print (menor_nota)
```

```python
# Exercício 7- Separando números positivos e negativos.

numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

print (numeros)
print (" ")

positivos = []
negativos = []

for numero in numeros:
    if numero >= 0:
        positivos.append(numero)
    else:
        negativos.append(numero)

print ("Números positivos: ")
print (positivos)
print (" ")
print ("Números negativos: ")
print (negativos)
```

```python
# Exercício 8- Ordenando uma lista.

numeros = [5, 2, 9, 1, 7, 3, 8, 4, 6, 10]

print ("Lista original: ")
print (numeros)
print (" ")
print ("Lista em ordem crescente: ")
numeros.sort()
print (numeros)
print (" ")
print ("Lista em ordem decrescente: ")
numeros.sort(reverse=True)
print (numeros)
print (" ")
print ("Maior valor e menor valor: ")
print (max(numeros))
(min(numeros))
```

```python
produtos = [
    ["Bala", 0.75],
    ["Pirulito", 1.0],
    ["Chocolate", 10.0],
    ["Cocada", 6.0],
    ["Chiclete", 2.0]
]

# Calcula o preço total somando apenas os valores (segundo elemento de cada sublista)
preco = sum(item[1] for item in produtos)

# Encontra o produto mais caro e mais barato usando o preço como critério
produto_mais_caro = max(produtos, key=lambda item: item[1])
produto_mais_barato = min(produtos, key=lambda item: item[1])

print ("Lista original: ")
print (produtos)
print (" ")
print ("Preço total da compra: ")
print (preco)
print (" ")
print ("Produto mais caro: ")
print (produto_mais_caro[0], "(", produto_mais_caro[1], ")")
print (" ")
print ("Produto mais barato: ")
print (produto_mais_barato[0], "(", produto_mais_barato[1], ")")
```

```python
# Exercício 10- Sistema de Notas.

num_alunos = int(input("Quantos alunos serão cadastrados? "))

alunos = []

for i in range(num_alunos):
    print(f"Aluno {i+1} ")
    nome = input("Digite o nome do aluno: ")
    while True:
        try:
            nota = float(input("Digite a nota do aluno (0-10): "))
            if 0 <= nota <= 10:
                break
            else:
                print("A nota deve estar entre 0 e 10. Tente novamente.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número para a nota.")

    alunos.append({'nome': nome, 'nota': nota})


print("Todos os alunos e suas notas:")
for aluno in alunos:
    print(f"- {aluno['nome']}: {aluno['nota']}")

total_notas = sum(aluno['nota'] for aluno in alunos)
media_turma = total_notas / num_alunos if num_alunos > 0 else 0
print(f"Média da turma: {media_turma:.2f}")

if alunos:
    aluno_maior_nota = max(alunos, key=lambda x: x['nota'])
    print(f"Aluno com a maior nota: {aluno_maior_nota['nome']} ({aluno_maior_nota['nota']})")

if alunos:
    aluno_menor_nota = min(alunos, key=lambda x: x['nota'])
    print(f"Aluno com a menor nota: {aluno_menor_nota['nome']} ({aluno_menor_nota['nota']})")

aprovados = [aluno for aluno in alunos if aluno['nota'] >= 7]
print("Alunos aprovados:")
if aprovados:
    for aluno in aprovados:
        print(f"- {aluno['nome']} ({aluno['nota']})")
else:
    print("Nenhum aluno aprovado.")

reprovados = [aluno for aluno in alunos if aluno['nota'] < 7]
print("Alunos reprovados:")
if reprovados:
    for aluno in reprovados:
        print(f"- {aluno['nome']} ({aluno['nota']})")
else:
    print("Nenhum aluno reprovado.")


while True:
    pesquisa_nome = input("Digite o nome do aluno para pesquisar (ou 'sair' para encerrar): ")
    if pesquisa_nome.lower() == 'sair':
        break

    encontrado = False
    for aluno in alunos:
        if aluno['nome'].lower() == pesquisa_nome.lower():
            print(f"A nota de {aluno['nome']} é {aluno['nota']}")
            encontrado = True
            break
    if not encontrado:
        print(f"Aluno '{pesquisa_nome}' não encontrado.")
```
